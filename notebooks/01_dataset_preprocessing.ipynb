## 3. Preprocessing

Langkah preprocessing untuk topik ini: resize seragam, konversi ruang warna ke LAB, dan pemisahan kanal luminans dan warna.

Sel kode di bawah berisi fungsi `praproses()` yang menjalankan langkah preprocessing kelompok. Alasan setiap langkah dan parameternya dijelaskan di sel markdown sesudahnya.


```python
import cv2
import numpy as np

UKURAN = (512, 512)


def praproses(citra):

    # 1. Resize
    citra_resize = cv2.resize(
        citra,
        UKURAN,
        interpolation=cv2.INTER_AREA
    )

    # 2. Konversi BGR -> LAB
    lab = cv2.cvtColor(
        citra_resize,
        cv2.COLOR_BGR2LAB
    )

    # 3. Pisahkan kanal L, A, dan B
    L, A, B = cv2.split(lab)

    # 4. Gabungkan kembali kanal LAB (tanpa modifikasi)
    lab_preprocessed = cv2.merge(
        [L, A, B]
    )

    # 5. Konversi LAB -> BGR
    hasil = cv2.cvtColor(
        lab_preprocessed,
        cv2.COLOR_LAB2BGR
    )

    return hasil


print("=" * 60)
print("PREPROCESSING CITRA MALAM")
print("=" * 60)
print(f"Ukuran standar citra : {UKURAN[0]} x {UKURAN[1]}")
print("Ruang warna          : BGR -> LAB")
print("Kanal yang digunakan : L, A, B")
print("Enhancement          : tidak dilakukan pada preprocessing")
print("Status               : Siap digunakan")
```

```python
# UJI PREPROCESSING PADA 1 CITRA

if not berkas:
    raise RuntimeError("Dataset citra tidak tersedia.")

# Ambil satu citra dari dataset
path_contoh = berkas[0]

# Baca citra
citra_asli = util.baca_citra(path_contoh)

if citra_asli is None:
    raise RuntimeError(
        f"Citra tidak dapat dibaca: {path_contoh}"
    )

# Jalankan preprocessing
citra_hasil = praproses(citra_asli)

# Tampilkan informasi hasil
print("=" * 60)
print("HASIL UJI PREPROCESSING")
print("=" * 60)

print(f"Nama file      : {path_contoh.name}")
print(
    f"Ukuran asli    : "
    f"{citra_asli.shape[1]} x {citra_asli.shape[0]}"
)
print(
    f"Ukuran hasil   : "
    f"{citra_hasil.shape[1]} x {citra_hasil.shape[0]}"
)
print(f"Tipe data hasil: {citra_hasil.dtype}")

print("Status         : Preprocessing berhasil diterapkan.")
```

**Alasan langkah preprocessing:**

Preprocessing dilakukan untuk menyeragamkan ukuran citra dan memisahkan informasi luminans dengan informasi warna sebelum citra diproses menggunakan metode perbaikan citra.

**1. Resize citra menjadi 512 × 512 piksel**

Seluruh citra diubah menjadi ukuran 512 × 512 piksel agar memiliki ukuran yang seragam. Ukuran yang seragam memudahkan proses pengolahan, perbandingan, dan evaluasi antar citra. Parameter yang digunakan adalah interpolasi `INTER_AREA`, yang sesuai untuk proses pengecilan ukuran citra.

**2. Konversi ruang warna BGR ke LAB**

Citra dikonversi dari ruang warna BGR ke ruang warna LAB agar informasi kecerahan dan warna dapat dipisahkan. Ruang warna LAB memiliki tiga kanal: kanal L merepresentasikan luminans (tingkat kecerahan), sedangkan kanal A dan B merepresentasikan informasi warna. Pemisahan ini sesuai dengan karakteristik citra malam yang umumnya memiliki permasalahan pada tingkat pencahayaan.

**3. Pemisahan kanal L, A, dan B**

Setelah dikonversi ke LAB, citra dipisahkan menjadi kanal L, A, dan B. Kanal L digunakan sebagai komponen luminans yang nantinya dapat diproses oleh metode peningkatan citra seperti CLAHE, sedangkan kanal A dan B dipertahankan sebagai informasi warna. Pada tahap preprocessing ini tidak dilakukan peningkatan kecerahan atau perubahan warna agar proses enhancement dapat dilakukan secara terpisah pada tahap metode klasik.

**Parameter preprocessing:**

- Ukuran citra: 512 × 512 piksel
- Interpolasi resize: `cv2.INTER_AREA`
- Konversi warna: BGR → LAB
- Kanal luminans: L
- Kanal warna: A dan B
- Enhancement: tidak dilakukan pada tahap preprocessing


## 4. Gambar sebelum–sesudah

Sel ini menyimpan 12 contoh ke folder `figures/` di Drive. Unggah minimal 10 ke `outputs/figures/` di repo (lihat `PANDUAN.md`).

```python
import random as _rng
FOLDER_FIGUR.mkdir(parents=True, exist_ok=True)

# ── Pisahkan foto sendiri vs LOL
_sendiri = [p for p in berkas if "LOL" not in str(p) and "terang" not in str(p)]
_lol     = [p for p in berkas if "LOL" in str(p) and "gelap" in str(p)]

# Seed tetap agar reproducible
_rnd = _rng.Random(2026)
_pilih_sendiri = _rnd.sample(_sendiri, min(6, len(_sendiri)))
_pilih_lol     = _rnd.sample(_lol,     min(6, len(_lol)))
_pilih = _pilih_sendiri + _pilih_lol

print(f"Foto sendiri dipilih : {len(_pilih_sendiri)} / {len(_sendiri)}")
print(f"Foto LOL dipilih     : {len(_pilih_lol)} / {len(_lol)}")
print(f"Total gambar figur   : {len(_pilih)}")
print()

# ── Simpan gambar sebelum–sesudah
for p in _pilih:
    asli = util.ubah_ukuran_maks(util.baca_citra(p), 1024)
    util.simpan_sebelum_sesudah(asli, praproses(asli), FOLDER_FIGUR / f"pra_{p.stem}.png")

contoh = sorted(FOLDER_FIGUR.glob("pra_*.png"))[:15]
print(len(list(FOLDER_FIGUR.glob("pra_*.png"))), "gambar di", FOLDER_FIGUR)
util.tampil(*[util.baca_citra(f) for f in contoh], kolom=1, ukuran=7)
```
